```python
import numpy as np 

class EvalPred:
    def __init__(self, preds, labels):
        self.elements = (preds, labels)

    def __iter__(self):
        return iter(self.elements)

    def __getitem__(self, idx):
        if idx < 0 or idx >= len(self.elements):
            raise IndexError("tuple index out of range")
        return self.elements[idx]
    
ep = EvalPred(np.array([1,0,1]), np.array([1,0,1]))
pred, true = ep
pred[0]
```

```python
# Load model directly
from transformers import AutoTokenizer, AutoModelForCausalLM

tokenizer = AutoTokenizer.from_pretrained("Qwen/Qwen2.5-0.5B")
model = AutoModelForCausalLM.from_pretrained("Qwen/Qwen2.5-0.5B")
messages = [
    {"role": "user", "content": "What is the capital of France?"},
]
inputs = tokenizer.apply_chat_template(
	messages,
	add_generation_prompt=True,
	tokenize=True,
	return_dict=True,
	return_tensors="pt",
).to(model.device)

outputs = model.generate(**inputs, max_new_tokens=40)
print(tokenizer.decode(outputs[0][inputs["input_ids"].shape[-1]:]))
```

```python
import numpy as np 
def circle(gridsize, cx, cy, radius):
    mat = np.zeros((gridsize, gridsize), dtype=int)
    ans = np.zeros((gridsize, gridsize), dtype=int)
    for x in range(1, gridsize - 1):
        for y in range(1, gridsize - 1):
            if (x - cx) ** 2 + (y - cy) ** 2 < radius ** 2:
                mat[x, y] = 1
    for x in range(gridsize):
        for y in range(gridsize):
            if (x > 0 and mat[x - 1, y] == 1) or (x < gridsize - 1 and mat[x + 1, y] == 1) or \
               (y > 0 and mat[x, y - 1] == 1) or (y < gridsize - 1 and mat[x, y + 1] == 1):
                if mat[x, y] == 0:
                    ans[x, y] = 1
    for x in range(gridsize):
        for y in range(gridsize):
            if ans[x, y] == 1:
                print('* ', end='')
            else:
                print('. ', end='')
        print()
```

```python
circle(20, 9, 9, 9)
```
